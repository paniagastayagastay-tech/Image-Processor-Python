```python
# Install Pillow (only needed in Colab)
!pip install pillow

# Import libraries
from PIL import Image, ImageFilter
from IPython.display import display

# Step 1: Upload image (for Google Colab)
from google.colab import files
uploaded = files.upload()

# Get uploaded file name
image_path = list(uploaded.keys())[0]

# Step 2: Open image
img = Image.open(image_path)

# Step 3: Resize image
resized = img.resize((200, 200))

# Step 4: Convert to grayscale
gray = resized.convert("L")

# Step 5: Apply blur filter
blur = resized.filter(ImageFilter.BLUR)

# Step 6: Display images
print("Original Image")
display(img)

print("Resized Image")
display(resized)

print("Grayscale Image")
display(gray)

print("Blur Image")
display(blur)

# Step 7: Save processed images
resized.save("resized_image.jpg")
gray.save("gray_image.jpg")
blur.save("blur_image.jpg")

# Step 8: Download saved images (Colab)
files.download("resized_image.jpg")
files.download("gray_image.jpg")
files.download("blur_image.jpg")
```
